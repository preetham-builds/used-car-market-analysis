Data analysis

```python
# Selling price details
df[df['selling_price'] == df['selling_price'].max()]
df['selling_price'].mean()
df['selling_price'].median()
```

```python
# Top 10 listed brands
df['brand'].value_counts().head(10)
```

```python
# Top 5 brands with highest avg selling price
df.groupby('brand')['selling_price'].mean().sort_values(ascending=False).head()
```

```python
# Which fuel type has the highest avg and median?
df.groupby('fuel_type')['selling_price'].mean().sort_values(ascending=False).head()
df.groupby('fuel_type')['selling_price'].median().sort_values(ascending=False).head()
```

```python
# Manual vs Automatic : How big is the price gap?
df.groupby('transmission_type')['selling_price'].mean().sort_values(ascending=True).diff().iloc[-1]
```

```python
# Price vs Vehicle age : How avg price changes year by year?
df.groupby('vehicle_age')['selling_price'].mean()
```

```python
# Individual vs Dealer
df.groupby('seller_type')['selling_price'].agg(['mean','median','count'])
```

```python
# Find the 10 most expensive diesel cars under 5 years old
df[(df['fuel_type']=='Diesel') & (df['vehicle_age']<5)].sort_values(by='selling_price',ascending=False).head(10)
```

```python
# Do engine size and power drive price?
df[['engine','max_power','mileage','vehicle_age','km_driven','selling_price']].corr()
# 'max_power' has a strong positive co-relation with 'selling_price'
```

```python
# What is the average price by fuel and transmission?
pd.pivot_table(df,index='brand',columns=['fuel_type','transmission_type'],values='selling_price',aggfunc='mean').head()
```

```python
# Group vehicle_age into multiple ranges
pd.cut(x=df['vehicle_age'],bins=[-1,5,10,20]).head()
```

```python
# Brands with more than 50 listings sorted by average selling price
df.groupby('brand').agg(count = ('selling_price','count'),avg_price = ('selling_price','mean')).query('count>50').sort_values('avg_price',ascending=False)
```

```python
q1,q3 = np.percentile(df['selling_price'],[25,75])
# The values are the selling prices below which the given % of values fall under
iqr = q3-q1
lower_bound = q1-1.5*iqr
upper_bound = q3+1.5*iqr
outliers = df[(df['selling_price']<lower_bound) | (df['selling_price']>upper_bound)]
outliers.head(10)
```

```python
df['log_price'] = np.log1p(df['selling_price'])
plt.figure(figsize=(12,5))
plt.subplot(1,2,1)
plt.hist(df['selling_price'],bins=30)
plt.title('Original Price')

plt.subplot(1,2,2)
plt.hist(df['log_price'],bins=30)
plt.title('Log price')

plt.show()
```

```python
median_price = df['selling_price'].median()
df['segment'] = np.where(df['selling_price']>median_price,'Premium','Budget')
df.groupby('segment')[['vehicle_age','km_per_year']].mean()
```

```python
df['km_per_year'].quantile([0.25,0.5,0.75])
conditions = [df['km_per_year']<5000,(df['km_per_year']>=5000) & (df['km_per_year']<=8000),df['km_per_year']>12000]
choices = ['Low','Medium','High']
df['usage'] = np.select(conditions,choices,default='Unknown')
df.groupby('usage')['selling_price'].mean()
```

```python
np.corrcoef(df['vehicle_age'],df['selling_price'])[0,1]
# Vehicle age and selling price has a moderately weak negative correlation. This means older vehicles tend to have lower prices
avg = df.groupby('vehicle_age')['selling_price'].mean()
change = np.diff(avg)/avg[:-1]*100
np.argmin(change) # Calculates the largest % drop in selling price and return the index of the smallest value
print("Highest depreciation : Year",avg.index[np.argmin(change)],'to Year',avg.index[np.argmin(change)+1])
```
